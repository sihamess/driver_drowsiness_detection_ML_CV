```python
import os
import time
from ultralytics import YOLO

# Specify the directory where your dataset.yaml is located and where the model weights are saved
dataset_path = '/content/gdrive/MyDrive/drowsiness_yolo/7.yolov8_yawdd/dataset.yaml'
weights_path = '/content/gdrive/MyDrive/drowsiness_yolo/7.yolov8_yawdd/runs/detect/train/weights/best.pt'

# Load the trained model
model = YOLO(weights_path)  # Load model with trained weights

# Evaluate the model to get the confusion matrix
t7 = time.time()
results = model.val(data=dataset_path, split='test') # Validate the model on the test set

t8 = time.time()
temps_evaluation = t8 - t7
print(f"Le temps d'évaluation est de {temps_evaluation} secondes.")
```
